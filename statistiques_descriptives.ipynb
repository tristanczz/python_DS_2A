```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd 

avg_creation = pd.read_csv(r"C:\Users\thoma\OneDrive\Documents\ENSAE\2A\Python-pour-la-data-science\Projet\python_DS_2A\data\Variables explicatives\avg_creation_2015_2025.csv")

```

```python
# 1. Définir le nom de votre colonne longue
col_valeur = "Création moyenne d'entreprise par an entre 2015 et 2025 par 1000 habitants"

# 2. Trier les données (Indispensable pour un barplot lisible !)
# On trie du plus grand au plus petit (ascending=False)
df_trie = avg_creation.sort_values(by=col_valeur, ascending=False)

# 3. Créer la figure (Taille 12x6 pour avoir de la place)
plt.figure(figsize=(12, 6))

# 4. Créer le Barplot
# palette='viridis' donne un dégradé de couleurs sympa
sns.barplot(data=df_trie, x='Region', y=col_valeur, palette='viridis')

# 5. Mettre en forme
plt.title("Taux de création d'entreprises par région (pour 1000 habitants)", fontsize=15, fontweight='bold')
plt.xlabel("Région")
plt.ylabel("Créations / 1000 hab.")

# Rotation des noms de régions pour qu'ils ne se chevauchent pas
plt.xticks(rotation=45, ha='right')

# Ajustement automatique pour ne rien couper
plt.tight_layout()

# 6. Afficher
plt.show()
```
