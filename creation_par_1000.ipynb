```python
import pandas as pd

creation_entreprise = pd.read_csv(r"C:/Users\thoma\OneDrive\Documents\ENSAE\2A\Python-pour-la-data-science\Projet\python_DS_2A\data\Variables explicatives\Variable_Y.csv")
creation_entreprise = creation_entreprise.rename(columns={'nom_region': 'region_nom'})

population_df = pd.read_csv(r"C:/Users\thoma\OneDrive\Documents\ENSAE\2A\Python-pour-la-data-science\Projet\python_DS_2A\data\Variables explicatives\population_par_region_annee.csv")
```

```python
creation_entreprise['region_nom'] = creation_entreprise['region_nom'].astype(str).str.replace("’", "'", regex=False)
creation_entreprise['TIME_PERIOD'] = pd.to_numeric(creation_entreprise['TIME_PERIOD'], errors='coerce')
```

```python
#Merge des deux dataframes sur la colonne 'region_nom'
merged_df = pd.merge(creation_entreprise, population_df, left_on='region_nom', right_on='region_nom', how='inner')
#creation d'entreprises par 1000 habitants
merged_df['creations_per_1000'] = (merged_df["nb_creations"] / merged_df['population']) * 1000
merged_df['creations_per_1000'] = merged_df['creations_per_1000'].round(2)
```

```python
#moyenne de créations par 1000 habitants par région
avg_creation = merged_df.groupby('region_nom', as_index=False)['creations_per_1000'].mean()
avg_creation = avg_creation.rename(columns={
'region_nom': 'Region',
'creations_per_1000': "Création moyenne d'entreprise par an entre 2015 et 2025 par 1000 habitants"
})
avg_creation["Création moyenne d'entreprise par an entre 2015 et 2025 par 1000 habitants"] = avg_creation["Création moyenne d'entreprise par an entre 2015 et 2025 par 1000 habitants"].round(2)

out_path = r"data/Variables explicatives/avg_creation_2015_2024.csv"
avg_creation.to_csv(out_path, index=False)
print(f"Exporté {len(avg_creation)} lignes vers {out_path}")

avg_creation.head()
```

```python
# Nom de la colonne longue (pour simplifier le code)
col_valeur = "Création moyenne d'entreprise par an entre 2015 et 2025 par 1000 habitants"

# 1. Trier
df_trie = avg_creation.sort_values(by=col_valeur, ascending=True)

# 2. Définir les colonnes qu'on veut voir
colonnes_a_afficher = ['Region', col_valeur]

# 3. Afficher les 3 premiers (avec la Région)
print("--- Les 3 plus faibles ---")
print(df_trie[colonnes_a_afficher].head(3))

# 4. Afficher les 3 derniers (avec la Région)
print("\n--- Les 3 plus élevés ---")
print(df_trie[colonnes_a_afficher].tail(3))
```
